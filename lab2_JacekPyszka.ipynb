```python
liczby = [
 3 + 4j,
 1 - 1j,
 -2j,
 np.exp(1j * np.pi / 3),
 2*np.exp(1j * np.pi / 4)
]

for z in liczby:
    print("z =", z)
    print("Re(z) =", z.real)
    print("Im(z) =", z.imag)
    print("sprzezenie =", np.conj(z))
    print("modul =", np.abs(z))
    print("faza [rad] =", np.angle(z))
    faza_stopnie = np.degrees(np.angle(z))
    print("faza [deg] =", faza_stopnie)
    print("kontrola z* z =", np.conj(z) * z)
    print("kontrola |z|^2 =", np.abs(z)**2)
    print()
```

### Zadanie 2 - Ket bra i iloczyn skalarny

```python
phi = np.array([
    [1/np.sqrt(2)],
    [1j/np.sqrt(2)]
    ], dtype=complex)

stany_psi = {
    "|0>": np.array([[1], [0]], dtype=complex),
    "|1>": np.array([[0], [1]], dtype=complex),
    "|+>": np.array([[1/np.sqrt(2)], [1/np.sqrt(2)]], dtype=complex),
    "|->": np.array([[1/np.sqrt(2)], [-1/np.sqrt(2)]], dtype=complex)
}

bra_phi = phi.conj().T
print("ket phi, ksztalt:", phi.shape)
print(phi)
print("bra phi, ksztalt:", bra_phi.shape)
print(bra_phi)
print()

for n, psi in stany_psi.items():
    iloczyn = (bra_phi @ psi).item()
    prawdopodobienstwo = abs(iloczyn)**2
    test_vdot = np.vdot(phi, psi)
    
    print(f"Stan psi: {n}")
    print(f"<phi|psi> = {iloczyn}")
    print(f"|<phi|psi>|^2 = {prawdopodobienstwo}")
    print(f"Test np.vdot: {test_vdot}\n")
```

Wniosek: Prawidłowa definicja wektora bra wymaga sprzężenia hermitowskiego, aby iloczyn skalarny wektora z samym sobą dawał rzeczywistą i nieujemną normę.

### Zadanie 3 - Normalizacja ortogonalność i projekcja

```python
def iloczyn_skalarny(phi, psi):
    return np.vdot(phi, psi)


def normalizuj(wektor):
    norma = np.sqrt(np.vdot(wektor, wektor).real)
    if np.isclose(norma, 0.0):
        raise ValueError("Wektor zerowy nie ma kierunku")
    return wektor/norma


def czy_ortogonalne(phi, psi, tol=1e-10):
    return np.isclose(iloczyn_skalarny(phi, psi), 0, atol=tol)

zero = np.array([1,0],dtype=complex)
jeden = np.array([0,1],dtype=complex)
plus = np.array([1/np.sqrt(2),1/np.sqrt(2)],dtype=complex)
minus = np.array([1/np.sqrt(2),-1/np.sqrt(2)],dtype=complex)

pary = {
    "|0>, |1>": (zero, jeden),
    "|+>, |->": (plus, minus),
    "|0>, |+>": (zero, plus)
}

v = np.array([1 + 1j, 2], dtype=complex)
v_norm = normalizuj(v)
prawd = np.abs(v_norm)**2

print("Znormalizowany v:", v_norm)
print("kontrola normy:", np.vdot(v_norm, v_norm))
print("Prawdopodobienstwo skladowych:", prawd[0],",", prawd[1])
print()

for nazwa, (phi,psi) in pary.items():
    iloczyn = iloczyn_skalarny(phi, psi)
    prawdopodobienstwo = np.abs(iloczyn)**2
    ortogonalnosc = czy_ortogonalne(phi, psi)
    print("Para:", nazwa)
    print("<phi|psi> =", iloczyn)
    print("prawdopodobienstwo projekcji =", prawdopodobienstwo)
    print("czy ortagonalne:", ortogonalnosc)
    print()
```

|Para|<phi psi>|abs(<phi psi>)**2|Ortogonalne|
|:-:|:-:|:-:|:-:|
| 0, 1 |  |  |  |
| +, - |  |  |  |
| 0, + |  |  |  |

Wniosek: Iloczyn skalarny to zespolona amplituda przechowująca informację o fazie, natomiast kwadrat jej modułu to rzeczywiste prawdopodobieństwo projekcji na dany stan.

### Zadanie 4 - Operatory projekcji

```python
zero = np.array([[1],[0]],dtype=complex)
one = np.array([[0],[1]],dtype=complex)

P0 = zero @ zero.conj().T
P1 = one @ one.conj().T

psi = normalizuj(np.array([[1+1j],[2]],dtype=complex))

skladowa_0 = P0 @ psi
skladowa_1 = P1 @ psi

p0 = (psi.conj().T @ P0 @ psi).item().real
p1 = (psi.conj().T @ P1 @ psi).item().real

print("P0 =",P0)
print("P1 =",P1)
print()
print("P0|psi> =",skladowa_0)
print("P1|psi> =",skladowa_1)
print()
print("p0, p1 =",p0, p1)
print("rekonstrukcja:", np.allclose(skladowa_0 + skladowa_1, psi))
print()
I = np.eye(2)
print("P0 + P1 == I:", np.allclose(P0 + P1, I))
print("P0^2 == P0:", np.allclose(P0 @ P0, P0))
print("P1^2 == P1:", np.allclose(P1 @ P1, P1))
```

Wniosek: Suma p0 + p1 = 1 wynika z faktu, że projektory wyczerpują wszystkie możliwe wyniki pomiaru, a zsumowanie działań obu operatorów na wektor pozwala na jego całkowitą, bezstratną rekonstrukcję.

### Zadanie 5 - Operatory macierzowe i kolejność działań

```python
A = np.array([[0, 1], [1, 0]], dtype=complex)
B = np.array([[1, 0], [0, -1]], dtype=complex)

zero = np.array([1, 0], dtype=complex)
one = np.array([0, 1], dtype=complex)

print("A|0> =", A @ zero)
print("A|1> =", A @ one)
print("AB =", A @ B)
print("BA =", B @ A)
print("AB = -BA:", np.allclose(A @ B, -(B @ A)))
print()

def czy_hermitowski(M):
    return np.allclose(M.conj().T, M)

def czy_unitarny(M):
    I = np.eye(M.shape[0])
    return np.allclose(M.conj().T @ M, I)

print("A hermitowski:", czy_hermitowski(A))
print("A unitarny:", czy_unitarny(A))
print("B hermitowski:", czy_hermitowski(B))
print("B unitarny:", czy_unitarny(B))
print()

print("B @ (A @ |0>)", B @ (A @ zero))
print("A @ (B @ |0>)", A @ (B @ zero))
print()

stan = np.array([1, 1j], dtype=complex) / np.sqrt(2)
norma_poczatkowa = np.linalg.norm(stan)
print("Sprawdzenie normy:")
print("Norma poczatkowa:", np.round(norma_poczatkowa, 4))
print("Norma po operacji A:", np.round(np.linalg.norm(A @ stan), 4))
print("Norma po operacji B:", np.round(np.linalg.norm(B @ stan), 4))
print()

wartosci, wektory = np.linalg.eigh(B)
print("Wartosci wlasne i wektory wlasne:")
print("Wartości własne:", wartosci)
print("Wektory własne (kolumny):", wektory)
```

Wniosek: Zapis BA|psi> czyta się od prawej do lewej, co wprost oznacza, że na wektor zadziała najpierw macierz A, a na uzyskany wynik nakładana jest macierz B.

### Zadanie 6 - Iloczyn tensorowy dwóch qubitów

```python
zero = np.array([1, 0], dtype=complex)
one = np.array([0, 1], dtype=complex)
plus = (zero + one) / np.sqrt(2)

baza = {
    "00": np.kron(zero, zero),
    "01": np.kron(zero, one),
    "10": np.kron(one, zero),
    "11": np.kron(one, one)
}

for etykieta, wektor in baza.items():
    print(etykieta, wektor, "dlugosc:", len(wektor))

print()
psi_2 = np.kron(plus, one)

print("stan dwoch qubitow:", psi_2)
print("prawdopodobienstwa:", np.abs(psi_2)**2)
print()

stan_qiskit = Statevector(psi_2)
print(stan_qiskit.probabilities_dict())
print("poprawny stan:", stan_qiskit.is_valid())

print("porownywanie kolejnosci:")
stan_A = np.kron(zero, plus)
stan_B = np.kron(plus, zero)

print("|0> tensor |+> =", stan_A)
print("|+> tensor |0> =", stan_B)
print("czy sa rowne?", np.array_equal(stan_A, stan_B))
```

Wniosek: Kolejność w iloczynie tensorowym przypisuje poszczególne stany do konkretnych, fizycznych qubitów układu. Odwrócenie kolejności reprezentuje inną konfigurację i tworzy zupełnie inny wektor.
