## Zadanie 1 - Amplitudy i reguła Borna

### Przykład do przepisania

```python
psi = np.array([
 np.sqrt(3) / 2, # amplituda stanu |0>
 1j / 2 # amplituda stanu |1>
], dtype=complex)
norma_kwadrat = np.vdot(psi, psi).real
prawdopodobienstwa = np.abs(psi)**2
print("Stan:", psi)
print("<psi|psi> =", norma_kwadrat)
print("P(0) =", prawdopodobienstwa[0])
print("P(1) =", prawdopodobienstwa[1])
print("Suma =", prawdopodobienstwa.sum())
```

### Stan |+>

```python
plus = np.array([
 1 / np.sqrt(2), # amplituda stanu |0>
 1 / np.sqrt(2) # amplituda stanu |1>
], dtype=complex)
norma_kwadrat = np.vdot(plus, plus).real
prawdopodobienstwa = np.abs(plus)**2
print("Stan:", plus)
print("<plus|plus> =", norma_kwadrat)
print("P(0) =", prawdopodobienstwa[0])
print("P(1) =", prawdopodobienstwa[1])
print("Suma =", prawdopodobienstwa.sum())
```

### Stan |->

```python
minus = np.array([
 1 / np.sqrt(2), # amplituda stanu |0>
 -1 / np.sqrt(2) # amplituda stanu |1>
], dtype=complex)
norma_kwadrat = np.vdot(minus, minus).real
prawdopodobienstwa = np.abs(minus)**2
print("Stan:", minus)
print("<minus|minus> =", norma_kwadrat)
print("P(0) =", prawdopodobienstwa[0])
print("P(1) =", prawdopodobienstwa[1])
print("Suma =", prawdopodobienstwa.sum())
```

#### Pomiar w bazie Z nie pozwala na odróżnienie stanu |+> od |->, ponieważ ich stany mają takie samo prawdopodobieństwo - 1/2.
#### Różnica między amplitudą a prawdopodobieństwem - amplituda to liczba zespolona określająca stan układu kwantowego i informację o fazie, a prawdopodobieństwo to liczba rzeczywista z przedziału [0, 1].

## Zadanie 2 - Normalizacja wektora stanu

### Uzupełnienie stan/norma, normalizacja [3, 4i]

```python
def normalizuj(stan):
    norma = np.sqrt(np.vdot(stan, stan).real)
    if np.isclose(norma, 0.0):
        raise ValueError("Nie mozna znormalizowac wektora zerowego")
    stan_znormalizowany = stan / norma
    return stan_znormalizowany, norma
phi = np.array([3, 4j], dtype=complex)
psi, norma = normalizuj(phi)
print("Przed normalizacja:", phi)
print("Po normalizacji:", psi)
print("Norma:", norma)
print("Kontrola:", np.vdot(psi, psi).real)
print("Prawdopodobienstwa:", np.abs(psi) ** 2)
```

## Zadanie 3 - Ten sam stan w Qiskit

### Kod startowy + analiza wektorów znormalizowanych z Zadania 2

```python
psi_numpy = np.array([
    np.sqrt(3) / 2,
    1j / 2
], dtype=complex)

psi_qiskit = Statevector(psi_numpy)

print(psi_qiskit)
print("Prawdopodobieństwa (Qiskit):", psi_qiskit.probabilities_dict())
print("Prawdopodobieństwa (Numpy):", np.abs(psi_numpy)**2)
print("Poprawny stan:", psi_qiskit.is_valid())

stany_znormalizowane = {
    "[3, 4i]": np.array([0.6+0j, 0+0.8j]),
    "[1, 1]": np.array([1/np.sqrt(2), 1/np.sqrt(2)]),
    "[1, -1]": np.array([1/np.sqrt(2), -1/np.sqrt(2)]),
    "[2 - 1i; -3i]": np.array([(2-1j)/np.sqrt(14), -3j/np.sqrt(14)])
}

for nazwa, wektor in stany_znormalizowane.items():
    sv = Statevector(wektor)
    print(f"Wektor {nazwa}: {wektor}")
    print("Prawdopodobieństwa (Qiskit):", sv.probabilities_dict())
    print("Prawdopodobieństwa (Numpy):", np.abs(wektor)**2)
    print("Poprawny stan:", sv.is_valid())
```

## Zadanie 4 - Wielokrotne pomiary i liczba prób

```python
def wykonaj_pomiary(stan, shots, seed=None):
    p = np.abs(stan) ** 2
    rng = np.random.default_rng(seed)
    wyniki = rng.choice([0, 1], size=shots, p=p)
    liczby = np.bincount(wyniki, minlength=2)
    return {
    "shots": shots,
    "zero": int(liczby[0]),
    "jeden": int(liczby[1]),
    "czestosc_0": liczby[0] / shots,
    "czestosc_1": liczby[1] / shots
}
psi = np.array([np.sqrt(3)/2, 1j/2], dtype=complex)
for shots in [10, 100, 1000, 10000]:
    wynik = wykonaj_pomiary(psi, shots, seed=42)
    # UZUPELNIJ: oblicz bezwzgledny blad czestosci jedynek
    blad = abs(wynik['czestosc_1']-0.25)
    print(f"Shots: {shots:5} | 0: {wynik['zero']:5} | 1: {wynik['jeden']:4} | Częstość 1: {wynik['czestosc_1']:.4f} | Błąd: {blad:.4f}")
```

```python
wynik = wykonaj_pomiary(psi, 10000, seed=42)
liczby = [wynik["zero"], wynik["jeden"]]
plt.bar(["0", "1"], liczby, color=["#1f77b4", "#ff7f0e"])
plt.xlabel("Wynik pomiaru")
plt.ylabel("Liczba wystapien")
plt.title("10000 symulowanych pomiarow")
plt.show()
```

### Zadanie 5 - Faza i interferencja

```python
def porownaj_drogi(A1, A2, opis):
    suma_klasyczna = np.abs(A1)**2 + np.abs(A2)**2
    wynik_interferencji = np.abs(A1 + A2)**2

    print(opis)
    print("|A1|^2 + |A2|^2 =", suma_klasyczna)
    print("|A1 + A2|^2 =", wynik_interferencji)
    print()

A1 = 1/np.sqrt(2)

porownaj_drogi(A1, 1/np.sqrt(2), "Ta sama faza")
porownaj_drogi(A1, -1/np.sqrt(2), "Przeciwna faza")
porownaj_drogi(A1, 1j/np.sqrt(2), "Roznica faz pi/2")
porownaj_drogi(A1, np.exp(1j*np.pi/3)/np.sqrt(2), "Czwarty przypadek")
```

### Zadanie 6 - Dwa kolejne pomiary

```python
def dwa_pomiary(stan, liczba_serii,seed=123):
    p = np.abs(stan)**2
    rng = np.random.default_rng(seed)
    pary = []

    for _ in range(liczba_serii):
        pierwszy = int(rng.choice([0, 1], p=p))
        drugi = pierwszy
        pary.append((pierwszy, drugi))
    
    return pary

plus = np.array([1, 1], dtype=complex) / np.sqrt(2)
for x in [20, 1000]:
    pary = dwa_pomiary(plus, x)
    print(pary)
    print("Czy wszystkie pary są zgodne?", all(a==b for a, b in pary))
```

W razie gdyby oba pomiary były niezależnie losowane z rozkładu 50/50, symulacja zignorowałaby zjawisko redukcji stanu kwantowego.
